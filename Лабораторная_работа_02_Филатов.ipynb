Необходимо вычислить значение выражения:
$$p=\frac{X}{1 \cdot 2 \cdot 3}+ \frac{Y^3}{3\cdot4\cdot5}-\frac{X^5}{5\cdot6\cdot7}+\frac{Y^7}{7\cdot8\cdot9}-...$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $X, Y$. Переменная, которая должна быть вычислена на основе независимых: $p$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для представления количества слагаемых.

Обратите внимание, что знаки слагаемых чередуются, степени числителя - нечетные числа по порядку, а множители в знаменателе - тройки чисел по порядку .

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $p$:

````python
X = 2
Y = 8
N = 10
p = 0
for n in range (1,N+1):
  chislo = (X if n % 2 != 0 else Y)**(2*n-1)
  znam = (2 * n - 1) * (2 * n) * (2 * n + 1)
  znak = -1 if n % 2 != 0 else 1
  p += znak * chislo / znam
  print('p =',p)
````
