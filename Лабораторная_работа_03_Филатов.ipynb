## Задание

Напишите функцию, которая принимает в качестве параметра одно
натуральное число N и печатает числа, принадлежащие отрезку [1;
N], кратные 3 или 5.

## Решение

### Объявление функции

```python
def chisla(N):
  a=[]
  for i in range(1,N+1):
    if i%3==0 or i%5==0:
      a.append(i)
  return a
```

### Демонстрация использования функции

```python
N=100
print('Числа от 1 до',N,', кратные 3 или 5:', chisla(N))
```
