## Задание

Напишите функцию, которая принимает в качестве параметра одно
натуральное число N и печатает числа, принадлежащие отрезку [1;
N], кратные 3 или 5.

## Решение

### Объявление функции

````python
def chisla(N):
  a=[]
  for i in range(1,N+1):
    if i%3==0 or i%5==0:
      a.append(i)
  return a
````

### Демонстрация использования функции

````python
N=100
print('Числа от 1 до',N,', кратные 3 или 5:', chisla(N))
````

1. **Функция с возвратом значения** возвращает результат через `return`, который можно использовать дальше. **Функция без возврата** выполняет действия (например, печать), но ничего не возвращает (по умолчанию `None`).

2. **Библиотечная функция** — готовая функция из стандартной или сторонней библиотеки (модуля), которую можно подключить и использовать, не писать самому.

3. **«Чёрный ящик»** — потому что для использования достаточно знать, что функция делает и какие аргументы принимает, а как она устроена внутри — не нужно.

4. **`x = random.randint(1, 100)`** — присваивает переменной `x` случайное целое число от 1 до 100 включительно.

5. **`print(random.randint(1, 20))`** — выводит случайное целое число от 1 до 20. (В записи опечатка: должны быть круглые скобки `()`, а не `{}`.)

6. **`random.seed(20)`** — задаёт начальное значение (зерно) генератора случайных чисел, чтобы последовательность «случайных» чисел была воспроизводимой.

7. **Повторное использование кода** — функцию пишут один раз, а вызывают многократно в разных местах программы (и в разных программах), не дублируя код.

8. **Две части определения функции:**
   - **Заголовок** — `def имя(параметры):` (имя, параметры).
   - **Тело** — блок инструкций с отступом, который выполняется при вызове.

9. **Параметры по умолчанию** задаются в заголовке через `=`:
```python
def greet(name='Гость'):
    print('Привет,', name)
```

10. **Рекурсия** — приём, когда функция вызывает саму себя (с условием выхода, чтобы не было бесконечности).
