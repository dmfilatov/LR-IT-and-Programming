## Задание

Необходимо вычислить значение выражения:
$$ee= \frac{g*h}{d17}+d17/h - \frac{g+d17+h}{4}$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $d17, g, h$. Переменная, которая должна быть вычислена на основе независимых: $ee$.

```python
d17=10
g=30.67
h=5
print("d17 = ", d17, ", type(d17) = ", type(d17))
print("g = ", g, ", type(g) = ", type(g))
print("h = ", h, ", type(h) = ", type(h))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $ee$:

```python
ee=((g*h)/d17)+(d17/h)-((g+d17+h)/4)
print('ee=',ee)
print('Type of ee: ', type(ee))
```
